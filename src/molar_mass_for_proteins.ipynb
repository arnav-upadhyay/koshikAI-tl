## Read in sequences

```julia
# Packages functions and data types --------
include("01_packages.jl") 

folder = "Sequences_welch_2009/"
sequence_name = "P1"
tags.growth_rate 	= 0.5
seq_file 			= folder * sequence_name * ".fasta" # Amino acid Sequence file in fasta format. Other option: DDC.fa

# Data import --------

## Import information about model species and sequences 
seqs, seq_names = species_and_sequences!(tags, seq_file)
seq = String(seqs[seq_names[1]].AA1[1:end-1])

# Data from https://proteomicsresource.washington.edu/protocols06/masses.php
# average masses
residue_mass = Dict(
		'A' => 71.0779000, # Alanin
		'R' => 156.1856800, # Arginin
		'N' => 114.1026400, # Asparagin
		'D' => 115.0874000, # Asparaginsäure
		'C' => 103.1429000, # Cystein
		'E' => 129.1139800, # Glutaminsäure
		'Q' => 128.1292200, # Glutamin
		'G' => 57.0513200, # Glycin
		'H' => 137.1392800, # Histidin
		'I' => 113.1576400, # Isoleucin
		'L' => 113.1576400, # Leucin
		'K' => 128.1722800, # Lysin
		'M' => 131.1960600, # Methionin
		'F' => 147.1738600, # Phenylalanin
		'P' => 97.1151800, # Prolin
		'S' => 87.0773000, # Serin
		'T' => 101.1038800, # Threonin
		'W' => 186.2099000, # Tryptophan
		'Y' => 163.1732600, # Tyrosin
		'V' => 99.1310600 # Valin
)
total = 0.0
for (i, aa) in enumerate(seq)
	if !haskey(residue_mass, aa)
		error("Unbekanntes Aminosäure-Symbol: '$aa' an Position $i")
	end
	total += residue_mass[aa]
end
# Abzug von Wasser für jede Peptidbindung (n-1)
n = length(seq)
mass_H2O = 18.01528
total -= (n - 1) * mass_H2O

println("Mass in Da: $total")

```
